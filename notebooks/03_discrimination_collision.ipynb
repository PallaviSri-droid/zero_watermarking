# Discrimination and collision analysis

Robustness alone is insufficient. We also test whether different images remain separated.

`collision_gap = min(inter-image HD) - max(intra-image HD)`

```python
import pandas as pd
from pathlib import Path
p=Path('experiments/results/benchmark_summary.csv')
assert p.exists(), 'Run notebook 02 first.'
df=pd.read_csv(p)
df[['method','mean_intra_hd','min_inter_hd','collision_gap','auc','eer']].sort_values('collision_gap',ascending=False)
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(figsize=(8,6))
ax.scatter(df.mean_intra_hd,df.mean_inter_hd,s=80)
for _,r in df.iterrows(): ax.annotate(r.method,(r.mean_intra_hd,r.mean_inter_hd),xytext=(4,4),textcoords='offset points',fontsize=8)
ax.set_xlabel('Intra-image HD ↓'); ax.set_ylabel('Inter-image HD ↑'); ax.set_title('Robustness vs discrimination'); plt.tight_layout()
```
