```python
import pandas as pd
from zero_watermarking.synthetic import make_dataset
from zero_watermarking.baselines import method_registry
from zero_watermarking.attacks import ATTACKS
from zero_watermarking.metrics import hamming
D=make_dataset(20,224); fn=method_registry(256)['KAZE-DCT-Balanced']
rows=[]
for aid,img in D.items():
    ref=fn(img)
    for attack,kwargs in [('gaussian_noise',{'sigma':s}) for s in [.01,.03,.05,.1]] + [('jpeg',{'quality':q}) for q in [90,70,50,30]]:
        z=fn(ATTACKS[attack](img,**kwargs)); rows.append({'id':aid,'attack':attack,'strength':str(kwargs),'hd':hamming(ref,z)})
r=pd.DataFrame(rows); r.groupby(['attack','strength']).hd.mean().reset_index()
```
