# Statistical confidence intervals

For the journal version, every headline metric should be reported with a bootstrap 95% confidence interval over image IDs or independent test cases. Never report a single seed as final evidence.

```python
import numpy as np

def bootstrap_ci(values, rounds=1000, seed=42):
    values=np.asarray(values,dtype=float)
    rng=np.random.default_rng(seed)
    means=[]
    for _ in range(rounds): means.append(rng.choice(values,size=len(values),replace=True).mean())
    return float(np.quantile(means,.025)), float(np.quantile(means,.975))

print('Use this helper on per-image/per-case metrics from the final test split.')
```
