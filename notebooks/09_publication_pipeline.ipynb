# 09 — Publication Benchmark Dashboard

Loads only generated result files and creates journal figures/tables. This notebook never invents values.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RESULTS=Path('../experiments/results')
FIG=Path('../figures/publication'); FIG.mkdir(parents=True,exist_ok=True)
summary=RESULTS/'journal_benchmark_summary.csv'
df=pd.read_csv(summary) if summary.exists() else pd.DataFrame()
df
```

```python
if not df.empty:
    ax=df.plot.bar(x='method',y='mean_nc',figsize=(11,5),legend=False)
    ax.set_ylabel('Mean NC'); ax.set_xlabel('Method'); ax.tick_params(axis='x',rotation=45)
    plt.tight_layout(); plt.savefig(FIG/'fig1_mean_nc.png',dpi=300); plt.show()
```

```python
if not df.empty and 'auc' in df:
    ax=df.plot.bar(x='method',y='auc',figsize=(11,5),legend=False)
    ax.set_ylabel('ROC-AUC'); ax.set_xlabel('Method'); ax.tick_params(axis='x',rotation=45)
    plt.tight_layout(); plt.savefig(FIG/'fig2_auc.png',dpi=300); plt.show()
```
